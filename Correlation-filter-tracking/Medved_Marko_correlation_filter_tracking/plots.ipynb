## Plots for the report

```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = {
    'Alpha': [0, 0.05, 0.1, 0.15, 0.2, 0.25, 0.3, 0.35, 0.4],
    'Average Overlap': [0.52, 0.52, 0.51, 0.50, 0.51, 0.52, 0.51, 0.50, 0.51],
    'Failures': [92, 58, 61, 56, 60, 60, 61, 66, 60]
}

df = pd.DataFrame(data)
display(df)
# Plot
fig, ax1 = plt.subplots(figsize=(12, 6))

color = 'tab:blue'
ax1.set_xlabel(r'$\alpha$', fontsize=18)

ax1.set_ylabel('Average Overlap', color=color, fontsize=18)
ax1.plot(df['Alpha'], df['Average Overlap'], marker='o', color=color) 
ax1.tick_params(axis='y', labelcolor=color, labelsize=14)
ax1.tick_params(axis='x', labelsize=14)

ax2 = ax1.twinx()
color = 'tab:red'
ax2.set_ylabel('Failures', color=color, fontsize=18)
ax2.plot(df['Alpha'], df['Failures'], marker='s', color=color)
ax2.tick_params(axis='y', labelcolor=color, labelsize=14)

fig.tight_layout()
plt.grid(True)
plt.savefig("report/figures/tracking_alpha.png")
```

```python
sigma_values = [0.75, 1.00, 1.25, 1.50, 1.75, 2.00, 2.25, 2.50, 2.75, 3.00]
average_overlap = [0.50, 0.50, 0.50, 0.52, 0.51, 0.51, 0.50, 0.50, 0.50, 0.49]
failures = [62, 61, 64, 58, 64, 62, 57, 62, 68, 68]


df = pd.DataFrame({
    'Sigma': sigma_values,
    'Average Overlap': average_overlap,
    'Failures': failures
})
display(df)

fig, ax1 = plt.subplots(figsize=(12, 6))

ax1.set_xlabel(r'$\sigma$', fontsize=18)
ax1.set_ylabel('Average Overlap', color='tab:blue', fontsize=18)
ax1.plot(df['Sigma'], df['Average Overlap'], color='tab:blue', label='Average Overlap', marker='o')
ax1.tick_params(axis='y', labelcolor='tab:blue', labelsize=14)
ax1.tick_params(axis='x', labelsize=14)


ax2 = ax1.twinx()
ax2.set_ylabel('Failures', color='tab:red', fontsize=18)
ax2.plot(df['Sigma'], df['Failures'], color='tab:red', label='Failures', marker='s')
ax2.tick_params(axis='y', labelcolor='tab:red', labelsize=14)

fig.tight_layout()
plt.grid(True)
plt.savefig("report/figures/tracking_sigma.png")
```

```python
# Time measurements 
data = {
    'Sequence': [
        'ball', 'basketball', 'bicycle', 'bolt', 'car', 'david', 'diving',
        'drunk', 'fernando', 'fish1', 'fish2', 'gymnastics', 'hand1', 'hand2',
        'jogging', 'motocross', 'polarbear', 'skating', 'sphere', 'sunshade',
        'surfing', 'torus', 'trellis', 'tunnel', 'woman'
    ],
    'Tracking Speed (FPS)': [
        1302.1, 379.4, 1877.0, 522.1, 1164.7, 930.9, 685.3,
        478.6, 188.8, 1373.6, 521.3, 634.2, 1248.8, 1348.1,
        1466.0, 179.9, 644.3, 615.1, 351.0, 1288.3,
        2157.4, 1388.3, 541.3, 632.6, 1020.5
    ]
}

df_speed = pd.DataFrame(data)
display(df_speed.sort_values(by="Tracking Speed (FPS)"))

df_sorted = df_speed.sort_values(by='Tracking Speed (FPS)', ascending=True)

plt.figure(figsize=(13, 10))
bars = plt.barh(df_sorted['Sequence'], df_sorted['Tracking Speed (FPS)'], color='skyblue')
plt.xlabel('Tracking Speed (FPS)', fontsize=18)
plt.ylabel('Sequence', fontsize=18)
plt.grid(axis='x', linestyle='--', alpha=0.6)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.xlim(0,2300)

for bar in bars:
    width = bar.get_width()
    plt.text(width + 20, bar.get_y() + bar.get_height()/2, f'{width:.1f}', va='center', fontsize=12)

plt.tight_layout()
plt.savefig("report/figures/tracking_speed.png")
```
