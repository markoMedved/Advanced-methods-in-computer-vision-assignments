```python
import numpy as np
import matplotlib.pyplot as plt
```

## Motion models and q

```python
q_values = np.array([0.001, 0.01, 0.1, 0.2, 0.4, 0.5, 0.6, 0.7, 0.8, 0.9, 1.0, 1.1, 1.2])

# Random walk 
average_overlap_RW = np.array([0.38, 0.40, 0.45, 0.48, 0.48, 0.49, 0.50, 0.50, 0.51, 0.50, 0.51, 0.50, 0.50])
total_failures_RW = np.array([150.0, 115.0, 63.0, 58.0, 44.0, 47.0, 38.0, 35.0, 35.0, 32.0, 31.0, 28.0, 30.0])
average_speed_RW = np.array([91.00, 87.51, 87.29, 86.19, 87.99, 85.66, 87.07, 92.08, 83.92, 87.84, 91.06, 87.32, 89.45])

# NCV
average_overlap_ncv = np.array([0.39, 0.42, 0.48, 0.50, 0.51, 0.50, 0.51, 0.52, 0.52, 0.50, 0.52, 0.52, 0.53])
total_failures_ncv = np.array([115.0, 95.0, 51.0, 46.0, 38.0, 35.0, 40.0, 36.0, 32.0, 29.0, 33.0, 31.0, 32.0])
average_speed_ncv = np.array([95.42, 93.71, 92.65, 91.14, 92.45, 93.83, 92.53, 89.45, 87.24, 89.82, 90.57, 87.86, 91.31])

# NCA
average_overlap_nca = np.array([0.41, 0.44, 0.46, 0.45, 0.46, 0.45, 0.46, 0.48, 0.46, 0.46, 0.47, 0.48, 0.46])
total_failures_nca = np.array([133.0, 116.0, 124.0, 109.0, 138.0, 136.0, 137.0, 152.0, 155.0, 144.0, 146.0, 168.0, 161.0])
average_speed_nca = np.array([89.62, 86.12, 80.90, 88.86, 94.62, 94.78, 89.30, 88.16, 87.57, 88.34, 84.71, 88.24, 83.46])
```

```python
fig, ax = plt.subplots(1,2, figsize=(25,10))

ax1 = ax[0]
ax1.plot(q_values, average_overlap_RW, '-o', label="RW")
ax1.plot(q_values, average_overlap_ncv, 'g-o', label="NCV")
ax1.plot(q_values, average_overlap_nca, '-o',color="orange", label="NCA")
ax1.set_xlabel("q", fontsize=30)
ax1.set_ylabel("Average Overalap", fontsize=30)
ax1.tick_params(axis="y")
ax1.legend( fontsize=30)

ax2 = ax[1]
ax2.plot(q_values, total_failures_RW, "-s", label="RW")
ax2.plot(q_values, total_failures_ncv, "-s", color="green", label="NCV")
ax2.plot(q_values, total_failures_nca, "-s", color="orange", label="NCA")
ax2.set_ylabel('Total Failures', fontsize=30)
ax2.set_xlabel("q", fontsize=30)
ax2.tick_params(axis='y')
ax2.legend(fontsize=30)
plt.savefig("report/figures/motion_models.png")
# NCA Probably larger overlap because it resets more times
```

## Impact of particles on tracking performance

```python
N_values = np.array([10, 30, 50, 75, 100, 125, 150])
average_overlap = np.array([ 0.47, 0.52, 0.51, 0.53, 0.50, 0.53, 0.53])
total_failures = np.array([ 82.0, 37.0, 34.0, 36.0, 29.0, 33.0, 31.0])
average_speed = np.array([ 677.51, 254.00, 161.42, 122.58, 86.64, 69.71, 62.55])
# NOTE That 1 particle was also tested but would ruin the plot
```

```python
fig,ax = plt.subplots(1,2, figsize=(25,10))
ax1 = ax[0]
ax1.plot(N_values, average_overlap, "-o", label = "Average Overlap")
ax1.set_ylabel("Average Overlap", color="#1f77b4", fontsize=30)
ax1.tick_params(axis="y", labelcolor="#1f77b4")
ax1.set_xlabel("Number of particles", fontsize=30)

ax2 = ax1.twinx()
ax2.plot(N_values, total_failures, "-s", color="orange", label = "Total Failures")
ax2.set_ylabel("Total Failures", color="orange", fontsize=30)
ax2.tick_params(axis="y", labelcolor="orange")
ax1.legend(loc = "center right", bbox_to_anchor = (1,0.42), fontsize=30)
ax2.legend(loc = "center right", bbox_to_anchor = (1,0.3), fontsize=30)

ax3 = ax[1]
ax3.plot(N_values, average_speed, "-o", color="grey")
ax3.set_ylabel("Average speed (FPS)", fontsize=30)
ax3.set_xlabel("Number of particles", fontsize=30)
plt.savefig("report/figures/particles.png")
```
